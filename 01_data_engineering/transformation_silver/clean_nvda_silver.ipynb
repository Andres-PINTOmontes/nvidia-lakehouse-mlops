```python
from pyspark.sql.functions import col, to_date

# 1. Leer de la tabla Bronze
df_bronze = spark.table("bronze_nvda_prices")

# 2. Selección y renombre de lo que SÍ existe
# Quitamos 'Adj Close' porque no está en la lista que imprimimos
df_silver = df_bronze.select(
    col("Date").alias("date"),
    col("Open").alias("open_price"),
    col("High").alias("high_price"),
    col("Low").alias("low_price"),
    col("Close").alias("close_price"),
    col("Volume").alias("volume")
)

# 3. Casteo de tipos: Asegurar que la fecha sea solo fecha y los precios sean decimales
df_silver = df_silver.withColumn("date", to_date(col("date")))

# 4. Mostrar para validar antes de guardar
display(df_silver.limit(5))
```

```python
# Guardamos en la Capa Silver
# Usamos 'overwrite' para que, si volvemos a correr el proceso, la tabla se actualice
df_silver.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_nvda_prices")

print("¡Tabla silver_nvda_prices creada y guardada!")
```
